Welcome to **Step 8E** of the Mint Leaf AI project (`13_robustness_statistical_validation.ipynb`). In this notebook, we perform a rigorous statistical validation of our 25-model benchmark experiment.

### 🔬 4 Core Statistical Analysis Objectives:
1. **95% Wilson Score Confidence Intervals**: Compute Wilson Score exact binomial confidence intervals for accuracy and per-class recall across minority classes (`Leaf_Spot` $N=8$, `Powdery_Mildew` $N=11$, `Mint_Rust` $N=14$).
2. **5-Fold Group-Aware Cross-Validation**: Evaluate stability across 5 independent group-aware folds to determine score variance ($	ext{Mean} \pm \text{Std}$).
3. **Pairwise McNemar Hypothesis Testing**: Test whether differences (e.g. 1 error vs 2 errors out of 313) are statistically significant ($p < 0.05$) or show no statistically significant superiority.
4. **Pareto Deployment Trade-off Analysis**: Map Macro F1 vs Checkpoint Size (MB) vs Inference Latency (ms) to identify production, edge, and baseline model roles.
