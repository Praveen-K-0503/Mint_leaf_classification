# 🌿 Mint Leaf AI — STEP 4: Disease Dataset Discovery & Acquisition Inventory

Welcome to **Step 4** of the Mint Leaf AI project. In this notebook (`03_disease_dataset_inventory.ipynb`), we conduct a factual, source-backed research audit of publicly available image datasets, open academic repositories, and agricultural extension archives containing genuine *Mentha* (mint) leaf disease images.

--- 

### 🎯 Objectives of Step 4:
1. **Public Dataset Inventory**: Identify open-access datasets on Kaggle, Roboflow Universe, Zenodo, Mendeley Data, GitHub, and USDA/CABI research archives.
2. **Strict Verification**: Ensure candidate datasets contain genuine *Mentha* species (distinguishing true mint from unrelated plants like *Plectranthus amboinicus* / Mexican Mint).
3. **Pathogen Coverage Analysis**: Evaluate data availability for key mint pathologies:
   - Mint Rust (*Puccinia menthae*)
   - Powdery Mildew (*Erysiphe cichoracearum*)
   - Septoria Leaf Spot (*Septoria menthae*)
   - Verticillium Wilt (*Verticillium dahliae*)
   - Rhizoctonia / Blight (*Rhizoctonia solani*)
   - Post-Harvest Spoilage / Decay
4. **No Downloading / Merging Yet**: We do not download or combine datasets in this step. This is purely an audit and inventory creation stage.
5. **Artifact Generation**: Export `mint_disease_dataset_inventory.csv`, `mint_disease_dataset_inventory.json`, and `mint_disease_sourcing_report.md` to `outputs/reports/dataset_sources/`.

---
